```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

url ="https://books.toscrape.com/index.html"

Response =requests.get(url)

soup = BeautifulSoup(Response.text,"html.parser")

books = soup.find_all("article", class_="product_pod")
print(books[0])
data = []

for book in books:
   title = book.find("h3").find("a")["title"]
   price = book.find("p",class_="price_color").text
   rating = book.find("p",class_="star-rating")["class"][1]

   data.append([title,price,rating])

df = pd.DataFrame(data,columns=["Title","Price","Rating"])
df.to_csv("books.csv",index=False)

print(df.head())

```

### Calculate Mean Price and Find Books at Mean Price

```python
lower_bound = 34.0
upper_bound = 40.0

books_in_custom_range = df[
    (df['Cleaned_Price'] >= lower_bound) &
    (df['Cleaned_Price'] <= upper_bound)
]

if not books_in_custom_range.empty:
    print(f"\nBooks priced between £{lower_bound:.2f} and £{upper_bound:.2f} (Mean Price: £{mean_price:.2f}):")
    display(books_in_custom_range[['Title', 'Price', 'Rating', 'Cleaned_Price']])
else:
    print(f"\nNo books found with prices between £{lower_bound:.2f} and £{upper_bound:.2f} (Mean Price: £{mean_price:.2f}).")
```

### Top-Rated Books (5-star and 4-star)

```python

top_rated_books = df[df['Rating'].isin(['Five', 'Four'])]

if not top_rated_books.empty:
    print("\nTop-rated books (5-star and 4-star):")
    rating_map = {'Five': 5, 'Four': 4, 'Three': 3, 'Two': 2, 'One': 1}
    top_rated_books['Numeric_Rating'] = top_rated_books['Rating'].map(rating_map)
    top_rated_books = top_rated_books.sort_values(by='Numeric_Rating', ascending=False).drop(columns=['Numeric_Rating'])

    display(top_rated_books[['Title', 'Price', 'Rating']])
else:
    print("\nNo books found with 5-star or 4-star ratings.")
```
