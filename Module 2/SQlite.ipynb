### Create Student Database

```python
def create_student_database(db_name='university.db', num_students=100):
    conn = ps.connect(db_name)
    Cursor = conn.cursor()

    Cursor.execute('''
        CREATE TABLE IF NOT EXISTS students (
            student_id INTEGER PRIMARY KEY AUTOINCREMENT,
            name_of_the_student TEXT,
            course_of_btech TEXT,
            cgpa REAL,
            email_id TEXT
        )
    ''')
    conn.commit()


    Cursor.execute('SELECT COUNT(*) FROM students')
    if Cursor.fetchone()[0] == 0:
        courses = ['Computer Science', 'Electrical Engineering', 'Mechanical Engineering', 'Civil Engineering']
        student_data = []
        for _ in range(num_students):
            name = fake.name()
            course = np.random.choice(courses)
            cgpa = round(np.random.uniform(6.0, 10.0), 2)
            email = fake.email()
            student_data.append((name, course, cgpa, email))

        Cursor.executemany("INSERT INTO students (name_of_the_student, course_of_btech, cgpa, email_id) VALUES (?, ?, ?, ?)", student_data)
        conn.commit()

    Cursor.execute("SELECT * FROM students")
    rows = Cursor.fetchall()
    df_students = pd.DataFrame(rows, columns=['student_id', 'name_of_the_student', 'course_of_btech', 'cgpa', 'email_id'])
    conn.close()
    return df_students

df_students = create_student_database()
print("Student Database created and populated (if empty):")
display(df_students.head())
```

### Create Faculty Database

```python
def create_faculty_database(db_name='university.db', num_faculty=3):
    conn = ps.connect(db_name)
    Cursor = conn.cursor()

    Cursor.execute('''
        CREATE TABLE IF NOT EXISTS faculty (
            faculty_id INTEGER PRIMARY KEY AUTOINCREMENT,
            name_of_faculty TEXT,
            subject TEXT,
            experience INTEGER,
            salary REAL
        )
    ''')
    conn.commit()

    Cursor.execute('SELECT COUNT(*) FROM faculty')
    if Cursor.fetchone()[0] == 0:

        subjects = ['Computer Science', 'Electrical Engineering', 'Mechanical Engineering']
        faculty_data = []
        for i in range(num_faculty):
            name = fake.name()
            subject = subjects[i]
            experience = np.random.randint(5, 25)
            salary = 0.0
            faculty_data.append((name, subject, experience, salary))

        Cursor.executemany("INSERT INTO faculty (name_of_faculty, subject, experience, salary) VALUES (?, ?, ?, ?)", faculty_data)
        conn.commit()

    Cursor.execute("SELECT * FROM faculty")
    rows = Cursor.fetchall()
    df_faculty = pd.DataFrame(rows, columns=['faculty_id', 'name_of_faculty', 'subject', 'experience', 'salary'])
    conn.close()
    return df_faculty

df_faculty = create_faculty_database()
print("Faculty Database created and populated (if empty):")
display(df_faculty.head())
```

### Calculate Faculty Salaries

```python
FEE_PER_STUDENT = 500

student_counts_by_subject = df_students.groupby('course_of_btech').size().reset_index(name='num_students')
student_counts_by_subject.rename(columns={'course_of_btech': 'subject'}, inplace=True)

df_faculty_with_counts = pd.merge(df_faculty, student_counts_by_subject, on='subject', how='left')

df_faculty_with_counts['num_students'] = df_faculty_with_counts['num_students'].fillna(0).astype(int)

df_faculty_with_counts['salary'] = df_faculty_with_counts['num_students'] * FEE_PER_STUDENT

print("Faculty salaries calculated based on student enrollment:")
display(df_faculty_with_counts)
```

### Join Student and Faculty Data

```python
df_students_renamed = df_students.rename(columns={'course_of_btech': 'subject'})

df_students_with_faculty = pd.merge(df_students_renamed,
                                    df_faculty_with_counts[['subject', 'name_of_faculty', 'faculty_id', 'salary']],
                                    on='subject',
                                    how='left')

print("Students mapped to their faculty:")
display(df_students_with_faculty.head())
```
